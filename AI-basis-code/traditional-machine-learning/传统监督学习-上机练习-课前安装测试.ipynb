```python
import numpy as np

# 随机采样两个高维高斯向量
d = 1000
u = np.random.randn(d)
v = np.random.randn(d)

# 计算并输出它们之间的距离与夹角
distance = np.linalg.norm(u - v)
angle = np.arccos(np.corrcoef(u, v)[0, 1]) / np.pi * 180

print(f"u 和 v 的距离为 {distance:.3f},  理论值 {(2 * d)**0.5:.3f}")
print(f"u 和 v 的夹角为 {angle:.3f}°, 理论值 90°")
```

```python
import matplotlib.pyplot as plt

# 带有噪声的模型 y = sin(x) + ϵ
x_sample = np.random.uniform(-6, +6, 100)
y_sample = np.sin(x_sample) + 0.1 * np.random.randn(100)

# 绘制散点图
plt.figure(figsize=(6, 4))
plt.scatter(x_sample, y_sample)
plt.xlabel("$X$", fontsize=14)
plt.ylabel("$Y$", fontsize=14)
plt.grid(alpha=0.5)
plt.show()
```

```python
import sklearn
from sklearn.linear_model import LinearRegression

# 生成简单含噪数据集
X = np.random.uniform(-5, 5, 100)
y = 2 * X - 3 +  + 0.1 * np.random.randn(100)

# 线性回归模型
model = LinearRegression()
model.fit(X.reshape(-1, 1), y)
k = model.coef_[0]
b = model.intercept_
print(f"线性回归结果 y = {k} * x {b:+}")

# 最小二乘法
A = np.array([[np.dot(X, X), np.sum(X)], [np.sum(X), X.shape[0]]])
c = np.array([np.dot(X, y), np.sum(y)])
k, b = np.linalg.solve(A, c)
print(f"最小二乘结果 y = {k} * x {b:+}")
```
